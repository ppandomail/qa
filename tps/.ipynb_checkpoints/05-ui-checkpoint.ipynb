# TP UI
---

Ejercicios de automatización de pruebas funcionales utilizando el sitio Coffee Cart (https://coffee-cart.app/)

1. **Verificación de carga inicial**. Desarrolle un caso de prueba automatizado que:
* Acceda a la página principal
* Verifique que se visualice el catálogo de bebidas
* Compruebe que el carrito de compras se encuentre inicialmente vacío

2. **Agregado de un producto al carrito**. Desarrolle una prueba automatizada que:
* Ingrese al sitio
* Seleccione una bebida cualquiera
* Verifique que la cantidad de productos en el carrito aumente en una unidad
* Verifique que el subtotal refleje correctamente la compra realizada

3. **Agregado de múltiples productos**. Automatice una prueba que:
* Agregue tres bebidas diferentes al carrito
* Verifique que la cantidad total de productos coincida con los artículos seleccionados
* Verifique que el importe total sea correcto

4. **Actualización dinámica del carrito**. Implementar una prueba que:
* Agregue un producto
* Capture el total actual
* Agregue un segundo producto
* Verifique que el importe total haya cambiado
* Verifique que la cantidad de productos se haya actualizado correctamente

5. **Compra completa**. Automatice el proceso de compra completo:
* Agregar al menos dos productos
* Acceder al carrito
* Iniciar el checkout
* Completar los datos solicitados
* Finalizar la compra
* Verificar la visualización del mensaje de confirmación

6. **Validación de campos obligatorios**. Construya una prueba automatizada que:
* Agregue un producto al carrito
* Inicie el proceso de compra
* Intente enviar el formulario sin completar datos
* Verifique que la aplicación informe errores de validación

7. **Eliminación de productos**. Desarrolle una prueba que:
* Agregue varios productos
* Elimine uno de ellos
* Verifique que la cantidad de productos disminuya
* Verifique la actualización del importe total

8. **Prueba basada en datos (Data Driven)**. Implemente una solución que ejecute la misma prueba utilizando distintos conjuntos de datos. Para cada ejecución:
* Agregar diferentes combinaciones de productos
* Calcular el importe esperado
* Comparar el valor mostrado por la aplicación con el valor calculado por la prueba

9. **Verificación de descuentos emergentes**. Analice el comportamiento promocional de la aplicación y diseñe una prueba automatizada que:
* Agregue productos en distintas cantidades
* Detecte la aparición de ofertas o descuentos
* Verifique que la promoción aplicada sea consistente con lo mostrado en pantalla

10. **Pruebas de regresión**. Construya una suite de regresión compuesta por:
* Carga de la aplicación
* Agregado de productos
* Actualización de cantidades
* Eliminación de productos
* Checkout exitoso

La suite deberá poder ejecutarse de forma automática mediante un único comando y generar un reporte HTML con los resultados obtenidos.

**Ejercicio Integrador Final**. Sistema: Coffee Cart

Desarrolle un framework de automatización que implemente como mínimo:

* 15 casos de prueba automatizados
* Patrón Page Object Model (POM)
* Uso de datos externos (CSV, JSON o Excel)
* Captura automática de screenshots ante fallos
* Generación de reportes HTML
* Ejecución desde línea de comandos
* Integración con GitHub Actions (opcional)

Entregables
* Código fuente
* Casos de prueba documentados
* Evidencias de ejecución
* Reporte generado automáticamente
* Informe técnico con conclusiones y defectos detectados

Este conjunto de ejercicios permite cubrir los conceptos más importantes de automatización web: localización de elementos, aserciones, sincronización, manejo de formularios, pruebas basadas en datos, Page Object Model y ejecución continua
