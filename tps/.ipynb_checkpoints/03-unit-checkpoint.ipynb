Imprimir por pantalla los números del 1 al 100 pero considerando lo siguiente: a) Si el número es divisible por 3 se debe imprimir "Fizz". b) Si el número es divisible por 5 se debe imprimir "Buzz". c) Si el número es divisible por 3 y por 5 se debe imprimir "FizzBuzz"

```python
def es_divisible(numero, divisor):
    return numero % divisor == 0

def fizzbuzz(numero):
    if es_divisible(numero, 3) and es_divisible(numero, 5):
        return 'FizzBuzz'
    if es_divisible(numero, 3):
        return 'Fizz'
    if es_divisible(numero, 5):
        return 'Buzz'
    return numero
    
def get_lista(numero):
    lista = []
    for i in range(1, numero+1):
        lista.append(fizzbuzz(i))  
    return lista
```

```python
import unittest

class FizzBuzzTest(unittest.TestCase):

    def test_fizz(self):
        self.assertEqual('Fizz', fizzbuzz(6))

    def test_buzz(self):
        self.assertEqual('Buzz', fizzbuzz(10))

    def test_fizzbuzz(self):
        self.assertEqual('FizzBuzz', fizzbuzz(15))

    def test_numero(self):
        self.assertEqual(7, fizzbuzz(7))

    def test_lista_hasta_15(self):
        esperado = [1, 2, 'Fizz', 4, 'Buzz', 'Fizz', 7, 8, 'Fizz', 'Buzz', 11, 'Fizz', 13, 14, 'FizzBuzz']
        self.assertEqual(esperado, get_lista(15))

unittest.main(defaultTest='FizzBuzzTest', argv=[''], verbosity=2, exit=False)
```

Los números expansivos se definen de la siguiente forma:
* 1 -> "un uno": 11
* 11 -> "dos unos": 21
* 21 -> "un dos, un uno": 1211
* 1211 -> "un uno, un dos, dos uno": 111221

Se pide programar: una función "expan" que, dado un arreglo con las cifras de un número, devuelva el número correspondiente

```python
def expandir(arr):
    resultado = []
    i = 0

    while i < len(arr):
        cifra = arr[i]
        contador = 1

        while i + 1 < len(arr) and arr[i + 1] == cifra:
            contador += 1
            i += 1

        resultado.append(str(contador))
        resultado.append(str(cifra))
        i += 1

    return int(''.join(resultado))
```

```python
class ExpanTest(unittest.TestCase):

    def test_expandir(self):
        casos = [([1], 11), ([1, 1], 21), ([2, 1], 1211), ([1, 2, 1, 1], 111221)]
        for valor, esperado in casos:
            with self.subTest(valor=valor):
                self.assertEqual(esperado, expandir(valor))

unittest.main(defaultTest='ExpanTest', argv=[''], verbosity=2, exit=False)
```
