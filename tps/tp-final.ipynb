# TP Integrador
---

## Testing Automatizado de Aplicaciones Reales

### Objetivo

Diseñar e implementar una estrategia de testing automatizado sobre uno o más sistemas existentes, utilizando:
* Tests unitarios
* Tests de API
* Tests web (UI)
* Generación de evidencias y reportes automáticos

### Testing Unitario (pytest)

1. Seleccionar un proyecto de código abierto desarrollado en Python
2. Analizar la funcionalidad del sistema
3. Identificar métodos críticos
4. Implementar al menos 5 tests unitarios: casos válidos, casos inválidos, valores límites, manejo de excepciones

### Testing de API (requests + pytest)

1. Realizar pruebas automáticas sobre una API REST pública. Sugerido: https://petstore.swagger.io
2. Implementar al menos 4 casos: post, get, put y delete
3. Validar: status code y datos

### Testing Web (selenium + pytest)

1. Realizar pruebas funcionales automatizadas sobre una aplicación web públcia. Sugerido: https://www.saucedemo.com
2. Casos:
    * credenciales válidas
    * contraseña incorrecta
    * visualización de productos
    * ordenamiento de precios
    * agregar producto al carrito
    * eliminar producto del pcarrito
    * verificar cantidad de productos en carrito
    * flujo de compra completo (validación de mensaje final)

### Framework de Automatización

El proyecto deberá presentar una estructura organizada:

```
tp_integrador/

├── unit_tests/
├── api_tests/
├── web_tests/
├── evidencias/
├── reportes/
├── requirements.txt
└── README.md
```

### Reportes

Generar automáticamente:
* Reporte HTML
* Evidencias de ejecución
* Capturas de pantalla de errores
* Resumen de pruebas ejecutadas

### Informe Final

El informe deberá incluir:
* Descripción de los sistemas seleccionados
* Estrategia de testing utilizada
* Casos de prueba diseñados
* Herramientas utilizadas
* Resultados obtenidos
* Incidencias encontradas
* Mejoras propuestas

### Desafio adicional

* Configurar un pipeline de CI/CD con GitHub Actions para que, ante cada push al repositorio, se ejecuten automáticamente los tests unitarios, de API y web, generando un reporte consolidado
* Este enfoque simula un escenario profesional de QA Automation y DevOps, donde se validan aplicaciones reales accesibles públicamente
