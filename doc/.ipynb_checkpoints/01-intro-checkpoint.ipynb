# QA automation
---

## Definición

* Es el uso de software especializado para ejecutar pruebas predefinidas en una aplicación sin intervención humana
* Este método compara los resultados reales con los esperados de forma rápida y repetitiva

## Objetivos

* Validar funcionalidades
* Detectar errores
* Garantizar el rendimiento del software de forma rápida y eficiente
* Fortalecer el Aseguramiento de Calidad durante todo el ciclo de desarrollo

## Beneficios Principales

* **Ahorro de tiempo**: ejecución rápida de tareas repetitivas y pruebas en paralelo
* **Mayor precisión**: menor cantidad de errores humanos al revisar procesos complejos
* **Cobertura constante**: ideal para pruebas de regresión cada vez que el código cambia
* **Importante para CI/CD**
* Permite que los tester se enfoquen en pensar escenarios nuevos y conocer más el negocio

## Tipos de Pruebas Automatizadas

* **Pruebas unitarias**: revisa pequeñas partes o funciones del código de manera aislada
* **Pruebas de integración**: verifica que distintos módulos funcionen bien juntos
* **Pruebas funcionales/UI/E2E**: simula las acciones de un usuario real (como clics y formularios) en la interfaz
* **Pruebas de APIs**: se enfoca en validar endpoints y servicios sin necesidad de interfaz gráfica
* **Pruebas de rendimiento**: analiza cómo se comporta el sistema bajo distintas condiciones
* **Pruebas de regresión**: se utiliza para asegurar que los cambios en el código no rompen funcionalidades existentes

## Funciones principales de un QA Tester

* Comprensión de los requisitos del producto o proyecto
* Diseño de estrategias de prueba eficientes
* Redacción de casos de prueba claros y escalables
* Ejecución de pruebas manuales y automatizadas
* Registro y documentación de bugs o anomalías detectadas

## ¿Qué tipos de herramientas necesita un proyecto de QA?

* Gestión de pruebas (Test Management)
* Automatización de pruebas
* Gestión de errores o bugs
* Integración con CI/CD y DevOps
* Visualización y reportes

## Herramientas de automatización de pruebas

* Selenium: Web (E2E)
* Playwright: Web (E2E)
* Cypress: Web (E2E)
* Postman: APIs
* Requests: APIs
* JMeter: Rendimiento, carga
