# Externalización de datos

* Externalizar los datos de entrada del test a un archivo
* El objetivo es desacoplar/deshardcodear los datos que consume la clase test, esto nos permite tener mayor legibilidad y hace mas simple el mantenimiento
* Instalar la libreria: pyaml_env

```python
import pytest
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from pyaml_env import parse_config

@pytest.mark.usefixtures('class_setup')
class TestPythonOrgSearch:

    datos = parse_config('data.yaml')

    @pytest.fixture
    def class_setup(self):
        self.driver = webdriver.Chrome()
        yield
        self.driver.close()

    def test_search_in_python_org(self):
        self.driver.get(self.datos['url'])
        assert 'Python' in self.driver.title
        elem = self.driver.find_element(By.NAME, 'q')
        elem.clear()
        elem.send_keys('pycon')
        elem.send_keys(Keys.RETURN)
        assert self.datos['expected'] not in self.driver.page_source
```

## Parametrize

* Permite ejecutar un mismo test con distintos datos
* Esto elimina el código repetitivo al generar automáticamente casos de prueba distintos y aislados para cada conjunto de entrada

### Ejemplo 1

```python
import pytest

@pytest.mark.parametrize("number", [1, 2, 3])
def test_is_positive(number):
    assert number > 0
```

### Ejemplo 2

```python
import pytest

def add(a, b):
    return a + b

@pytest.mark.parametrize("a, b, expected", [
    (1, 2, 3),
    (5, 5, 10),
    (-1, 1, 0),
])
def test_add(a, b, expected):
    assert add(a, b) == expected
```
