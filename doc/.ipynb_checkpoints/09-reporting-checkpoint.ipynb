# Reporting
---

* Es el proceso de documentar, analizar y presentar los resultados de las pruebas de software
* Ofrece una visión clara del estado de calidad del sistema a los desarrolladores, gestores y otros implicados para tomar decisiones informadas sobre el lanzamiento

## Componentes de un informe de pruebas

* **Resumen de ejecución**: muestra cuántas pruebas pasaron, fallaron o quedaron bloqueadas
* **Métricas de defectos**: detalla la cantidad, gravedad y distribución de los errores encontrados
* **Cobertura**: indica qué porcentaje de código o requisitos funcionales se logró evaluar
* **Conclusiones y riesgos**: evalúa si el producto cumple con los criterios de calidad necesarios para salir al mercado

## Allure report

### 1. Configuración e instalación

```sh
pip install allure-pytest
```

### 2. Anotar los tests con metadatos

````python
import pytest
import allure

@allure.epic('Autenticación')
@allure.feature('Login de Usuarios')
@allure.story('Login Exitoso con Credenciales Válidas')
@allure.severity(allure.severity_level.CRITICAL)
def test_login_exitoso():
    '''Prueba que un usuario con credenciales correctas pueda ingresar al sistema.'''
    
    with allure.step('Paso 1: Abrir la página de inicio de sesión'):
        # Código para abrir el navegador
        url_actual = 'https://ejemplo.com'
    
    with allure.step('Paso 2: Ingresar usuario y contraseña válidos'):
        usuario = 'admin'
        contrasena = 'password123'
        allure.attach(f'Usuario: {usuario}', name='Datos de Entrada', attachment_type=allure.attachment_type.TEXT)

    with allure.step("Paso 3: Hacer clic en el botón 'Iniciar Sesión'"):
        login_exitoso = True  # Simulación de la acción

    with allure.step('Paso 4: Verificar redirección al panel principal'):
        assert login_exitoso is True, 'El usuario no pudo iniciar sesión de forma correcta'
````

### 3. Ejecución y Generación del Reporte

```sh
pytest --alluredir=allure-results
allure serve allure-results
```

## Estructura Visual del Reporte Generado

* Cuando se ejecuta el comando allure serve, se abre una interfaz web interactiva
* Conceptualmente, la distribución del Dashboard Principal se organiza de la siguiente manera:
* El gráfico de torta (Test Suites Status) agrupa visualmente los resultados utilizando un código de colores estándar:
    * 🟢 Passed: pruebas exitosas
    * 🔴 Failed: errores de aserción (fallas en el producto)
    * 🟡 Broken: errores en el código de la prueba o configuración de la infraestructura
    * ⚪ Skipped: pruebas omitidas intencionalmente
